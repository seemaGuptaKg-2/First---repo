# python implementation of simple Linear Regression on salary data of software engineers

# step 1: import the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

# step 2: reading the data and splitting it to input and output


```python
dataset = pd.read_csv('https://raw.githubusercontent.com/enuguru/aiandml/master/datasets/salary_regression_train.csv')
inputx = dataset.iloc[:, :-1].values
outputy = dataset.iloc[:, 1:].values
print(outputy)
```

```python
print(dataset)
```

# step 3: select one thirds of the data for testing and two thirds for training

```python
input_train, input_test, output_train, output_test = train_test_split(inputx, outputy, test_size = 1/4, random_state = 7)
print("i/p train: ", input_train)
print("i/p test: ", input_test)
print("o/p train: ", output_train)
print("o/p test: ", output_test)
```

# step 4: selecting the simple Linear Regression model

```python
model = LinearRegression()
print("\nThe parameters of the model are: ",model.get_params())
#print(model.set_params())
print("\nThe model we are using is: ", model.fit(input_train, output_train))
```

# step 5: testing or model prediction using testinput

```python
years = float(input("\nGive number of years of experience  "))
testinput = [[years]]
predicted_output = model.predict(testinput)
print('\nThe number of years of experience is ',testinput)
print('\nThe salary for the number of years of experience is ',predicted_output)
yes = input("\nCan I proceed\n")
```

# step 6: Visualising the training results

```python
plt.scatter(input_train, output_train, color = 'Teal')
plt.plot(input_train, model.predict(input_train), color = 'orange')
plt.title('Salary vs Experience (Training set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

# step 7: Printing the testing results

```python
print("\nThe test input (number of years of experience) is as follows \n")
print(input_test)
# model predicting the Test set results
predicted_output = model.predict(input_test)
print("\nThe output (salary) for the test input is as follows \n")
print(predicted_output)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler

dataset1 = pd.read_csv('https://raw.githubusercontent.com/seemaGuptaKg-2/My-data/refs/heads/main/Attirition_data.csv')
print("Dataset1: ", dataset1)
label_encoder = LabelEncoder()
dataset1['JobRole'] = label_encoder.fit_transform(dataset1['JobRole'])
input_x = dataset1.iloc[:,:-1].values
output_y = dataset1.iloc[:,5].values

input_train, input_test, output_train, output_test = train_test_split(input_x, output_y, test_size = 1/5, random_state = 11)

# Define the feature names (column names)
feature_names = ['Age', 'JobRole', 'MonthlyIncome', 'JobSatisfaction', 'YearsAtCompany']

# Convert the NumPy array to a Pandas DataFrame with specified column names
input_train = pd.DataFrame(input_train, columns=feature_names)

# Create and train the linear regression model
model = KNeighborsClassifier(n_neighbors=3)
model.fit(input_train,output_train)

output_test = model.predict(input_test)
print("input_test: ", input_test)
print("output_test: ", output_test)

#predict for given case as input
print("Enter new employee details to predict attrition:")
inputx_age = input("Age: ")
inputx_JobRole = input("JobRole: ")
print("inputx_jobRole")
inputx_JobRole = label_encoder.transform([inputx_JobRole])[0]
print("inputx_jobRole: ",inputx_JobRole)
inputx_MonthlyIncome = input("MonthlyIncome: ")
inputx_JobSatisfaction = input("JobSatisfaction: ")
inputx_YearsAtCompany = input("YearsAtCompany: ")

testinput = pd.DataFrame([[inputx_age,inputx_JobRole,inputx_MonthlyIncome,inputx_JobSatisfaction,inputx_YearsAtCompany]],columns=['Age','JobRole','MonthlyIncome','JobSatisfaction','YearsAtCompany'])
print(testinput)

testoutput = model.predict(testinput)
print("\nPrediction (1 = Attrition, 0 = No Attrition):", testoutput[0])

print(input("Press any key to continue?"))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # Enables 3D plotting
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

dataset1 = pd.read_csv('https://raw.githubusercontent.com/seemaGuptaKg-2/My-data/refs/heads/main/housing_dataset.csv')
print(dataset1)

x=dataset1.iloc[:,:-1].values
y=dataset1.iloc[:,2].values

# Create and train the linear regression model
model = LinearRegression()
model.fit(x, y)
price_Pred = model.predict(x)

fig = plt.figure(figsize=(8,6))   # Create figure of size 8x6 inches
ax = fig.add_subplot(111, projection='3d')  # 111 → 1 row, 1 col, first plot
ax.scatter(x[:,0], x[:,1], y, c='r', marker='o')  # Plot red scatter points


# Plot grid
area = np.linspace(x[:,0].min(), x[:,0].max(), 100)
bedroomcount = np.linspace(x[:,1].min(), x[:,1].max(), 100)
area, bedroomcount = np.meshgrid(area, bedroomcount)

#Flatten the 2 axes
grid = np.array([area.flatten(), bedroomcount.flatten()]).T
price_Pred = model.predict(grid).reshape(area.shape)

ax.plot_surface(area, bedroomcount, price_Pred, cmap="jet")  # Colorful surface
plt.show()
```
